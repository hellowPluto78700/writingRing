## Interpretation guide

- Joint ↑ probes with small native gain → representation organization improved but time-shared readout still limits task BA.
- MM ↑ L1/L2 spike probes → heterogeneous thresholds preserve binary communication information.
- L1-TSCE ↑ L1 spike probe but not L2/native → local supervision works but downstream transform/readout does not exploit it.
- Positive MT×Joint or MT×L1-TSCE → richer learned representation benefits specifically from richer binary communication.
- If hidden probes improve while native BA stays low, phase-aware readout remains a separate next-stage question.
