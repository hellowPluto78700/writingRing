## Interpretation guide

- L2-memory benefit: native/L2 communication improves as L2 shift increases.
- Weighted recovery: communication Fixed250 improves while pre-reset BA is preserved, so communication delta becomes less negative.
- Complementarity: positive coding×memory interaction.
- Redundancy: Weighted31 removes the benefit of longer L2 memory.
- Burst/over-integration: weighted gains coincide with high count>=2/4 or cap-hit rates and poorer native BA.
- Temporal specialization: Fixed250 improves while whole/native does not.
